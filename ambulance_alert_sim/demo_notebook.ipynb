```python
from simulator import (
    get_base_density,
    simulate_no_alert,
    simulate_with_alert,
)
from config import H_default, alpha_default
import matplotlib.pyplot as plt
```

```python
base_density = get_base_density()
print("Base density:", base_density)
```

```python
positions_no, time_no = simulate_no_alert(base_density)
positions_alert, time_alert = simulate_with_alert(
    base_density, H=H_default, alpha=alpha_default
)
print("No alert:", time_no)
print("With alert:", time_alert)
print("Time saved:", time_no - time_alert)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(range(len(positions_no)), positions_no, label="No alert")
plt.plot(range(len(positions_alert)), positions_alert,
         label="With advance alert")
plt.xlabel("Time step")
plt.ylabel("Ambulance position (segment)")
plt.title("Ambulance movement comparison")
plt.legend()
plt.grid(True)
plt.show()
```
